# Analyse statistique — rendements × climat

Chemin fiable (sans Jupyter) : `python -m agri_climat analyse`

La logique vit dans `src/agri_climat/analyse.py`. **Pas de `plt.show()`**.

```python
from agri_climat.analyse import (
    add_yield_residuals,
    atypical_years,
    correlations_by_crop,
    plot_spearman_heatmap,
    plot_wheat_deep_dive,
    sensitivity_ranking,
)
from agri_climat.eda_preview import load_joined_table

joined = load_joined_table()
table = add_yield_residuals(joined)
corr = correlations_by_crop(table)
sensitivity_ranking(corr)
```

## Années atypiques (Wallonie)

Résidu de rendement z ≤ −1 **et** au moins un |z| climatique de saison ≥ 1.

```python
atypical_years(table)
```
