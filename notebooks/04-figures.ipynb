# Figures README — rendements × climat

Chemin fiable (sans Jupyter) : `python -m agri_climat figures`

La logique vit dans `src/agri_climat/figures.py`. **Pas de `plt.show()`.**

```python
from agri_climat.analyse import (
    add_yield_residuals,
    atypical_years,
    correlations_by_crop,
    sensitivity_ranking,
)
from agri_climat.eda_preview import load_joined_table
from agri_climat.figures import (
    plot_at_risk_years,
    plot_residual_vs_climate,
    plot_sensitivity_ranking,
)

joined = load_joined_table()
table = add_yield_residuals(joined)
ranking = sensitivity_ranking(correlations_by_crop(table))
ranking
```

## Trois figures du README

Les PNG sont écrits par le CLI dans `pictures/readme/`. Ici on construit les mêmes figures sans ouvrir de fenêtre Tk.

```python
import matplotlib

matplotlib.use("Agg")

fig_rank = plot_sensitivity_ranking(ranking)
fig_scatter = plot_residual_vs_climate(table, ranking)
fig_years = plot_at_risk_years(table, atypical_years(table))
print(len(fig_rank.axes), "axe ranking")
print(sum(ax.get_visible() for ax in fig_scatter.axes), "facettes nuages")
print(len(fig_years.axes), "panneaux années à risque")
```
