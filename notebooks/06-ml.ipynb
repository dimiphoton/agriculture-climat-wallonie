# ML basique — leave-one-year-out

Chemin fiable (sans Jupyter) : `python -m agri_climat ml`

La logique vit dans `src/agri_climat/ml.py`. **Pas de `plt.show()`.**

```python
from agri_climat.analyse import add_yield_residuals
from agri_climat.eda_preview import load_joined_table
from agri_climat.ml import evaluate_all_crops, wheat_loo_detail

table = add_yield_residuals(load_joined_table())
evaluate_all_crops(table)
```

## Froment : observé vs prédiction leave-one-out

```python
import matplotlib

matplotlib.use("Agg")

from agri_climat.ml import plot_mae_vs_naive, plot_wheat_loo

metrics = evaluate_all_crops(table)
detail = wheat_loo_detail(table)
fig_m = plot_mae_vs_naive(metrics)
fig_w = plot_wheat_loo(detail)
print(len(fig_m.axes), "axe MAE ;", len(fig_w.axes), "panneaux froment")
```
